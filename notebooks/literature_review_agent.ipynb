# HackBio literature review agent

Search PubMed, fetch abstracts, and ask Claude for a cited literature review.

Set `ANTHROPIC_API_KEY` in the repo's `.env` file (copy `.env.example`) before running the review cell.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))

import pandas as pd
from IPython.display import display, Markdown

import literature_agent as agent
print('Model:', agent.MODEL, '| effort:', agent.EFFORT, '| NCBI email:', agent.NCBI_EMAIL)
```

## 1. Ask a question

```python
QUESTION = "What is known about the role of TP53 mutations in glioblastoma treatment resistance?"
QUERY = "TP53[tiab] AND glioblastoma[tiab] AND resistance[tiab]"  # or None to search with the question
MAX_RESULTS = 10
```

## 2. Search PubMed and fetch abstracts (no API key needed)

```python
pmids = agent.search_pubmed(QUERY or QUESTION, max_results=MAX_RESULTS)
articles = agent.fetch_articles(pmids)
pd.DataFrame([{'pmid': a.pmid, 'year': a.year, 'journal': a.journal, 'title': a.title} for a in articles])
```

## 3. Ask Claude for the review

```python
review = agent.review_literature(QUESTION, articles)
display(Markdown(review))
```
